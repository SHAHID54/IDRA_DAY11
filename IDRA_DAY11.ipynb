```python
# To remove extra spaces from Gender values
df["Gender"] = df["Gender"].str.strip()
```

```python
# To standardize Gender values
df["Gender"] = df["Gender"].str.title()
```

```python
# To display the Gender values
df["Gender"].value_counts(dropna=False)
```

```python
# To standardize City values
df["City"] = df["City"].replace({
    "delhi": "Delhi",
    "DELHI": "Delhi"
})
```

```python
# To convert Joining Date into datetime format
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"], errors="coerce")
```

```python
# errors="coerce": Replaces invalid or broken date entries with NaN / NaT
```

```python
# To check the Joining Date data type
df["Joining_Date"].dtype
```

```python
# To remove duplicate rows
df = df.drop_duplicates()

# To check the number of rows after removing duplicates
df.shape
```

```python
# From shape we come to know before there were 157 rows , now they became 150, 7 rows were removed
```

```python
# To check the missing values after removing duplicates
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
# To calculate the median Age
df["Age"].median()
```

```python
# we already know median = mid value
```

```python
# To fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# To check missing Age values
df["Age"].isnull().sum()
```

```python
# here we filed all missing values with median of age column
```

```python
# To calculate the median Annual Salary
df["Annual_Salary"].median()
```

```python
# To fill missing Annual Salary values with the median
df["Annual_Salary"] = df["Annual_Salary"].fillna(df["Annual_Salary"].median())

# To check missing Annual Salary values
df["Annual_Salary"].isnull().sum()
```

```python
# To calculate the median Experience
df["Experience_Years"].median()
```

```python
# To fill missing Experience values with the median
df["Experience_Years"] = df["Experience_Years"].fillna(df["Experience_Years"].median())

# To check missing Experience values
df["Experience_Years"].isnull().sum()
```

```python
# To calculate the median Performance Score
df["Performance_Score"].median()
```

```python
# To fill missing Performance Score values with the median
df["Performance_Score"] = df["Performance_Score"].fillna(df["Performance_Score"].median())

# To check missing Performance Score values
df["Performance_Score"].isnull().sum()
```

```python
# To calculate the mode of Department
df["Department"].mode()
```

```python
# To fill missing Department values with the mode
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

# To check missing Department values
df["Department"].isnull().sum()
```

```python
# To calculate the mode of Gender
df["Gender"].mode()
```

```python
# To fill missing Gender values with the mode
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

# To check missing Gender values
df["Gender"].isnull().sum()
```

```python
# To calculate the mode of Work Mode
df["Work_Mode"].mode()
```

```python
# To fill missing Work Mode values with the mode
df["Work_Mode"] = df["Work_Mode"].fillna(df["Work_Mode"].mode()[0])

# To check missing Work Mode values
df["Work_Mode"].isnull().sum()
```

```python
# To check all remaining missing values
df.isnull().sum()
```

```python
# To calculate the mode of City
df["City"].mode()
```

```python
# To fill missing City values with the mode
df["City"] = df["City"].fillna(df["City"].mode()[0])

# To check missing City values
df["City"].isnull().sum()
```
